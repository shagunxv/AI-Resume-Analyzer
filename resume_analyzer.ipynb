```python
from google.colab import files
uploaded= files.upload() #upload resume
```

```python
from PyPDF2 import PdfReader
pdf_file=list(uploaded.keys())[0]
reader = PdfReader(pdf_file)
text=""
for page in reader.pages:
  text+=page.extract_text()
print(text) #PDF text extracted
```

```python
skills=[
    "Python",
    "Machine Learning",
    "C",
    "Pandas",
    "NumPy",
    "Data Analysis",
    "Data Cleaning",
    "EDA",
    "Feature Engineering",
    "Scikit-learn",
    "HTML",
    "CSS",
    "JavaScript",
    "DJango",
    "Neural Networks"
]
found_skills=[]
for skill in skills:
  if skill in text:
    found_skills.append(skill)
print("Skills Found: ")
print(found_skills) #Skill extraction from PDF
```

```python
job_desc="""
Looking for Pyhton, C, Machine Learning, DJango, Neural Networks and Data Analysis skills.
"""
required_skills=[]
for skill in skills:
  if skill in job_desc:
    required_skills.append(skill)
matched=set(found_skills) & set(required_skills)
score=len(matched)/len(required_skills)*100
print(f"ATS Match Score: {score:2f}%") #Showing ATS score.
```

```python
missing_skills=set(required_skills)-set(found_skills)
print("Missing Skills: ")
print(missing_skills)
```

```python
import matplotlib.pyplot as plt
labels=['Matched','Missing']
sizes=[len(matched),len(missing_skills)]
plt.pie(sizes,labels=labels,autopct='%1.1f%%')
plt.title('ATS Match Score')
plt.show() #visualization
```
